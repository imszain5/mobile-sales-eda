## Section 3: Clean the data
Fix missing values, data types, and duplicates.

**Hints from Week 1:**
- Find missing: `df.isnull().sum()`
- Fill numeric with median / 0 where it makes sense; fill categories with mode
- Drop rows only when you cannot reasonably fill (e.g. Customer Name)
- Convert dates: `pd.to_datetime(..., format='mixed')` if formats are mixed
- Remove duplicates: `df.drop_duplicates()` (remember to reassign to `df`)

### 3.1 Find missing values

```python
# Count missing values per column
df.isnull().sum()
```

### 3.2 Handle missing values
Decide a strategy for each column that has missing data.  
Write a short comment above each fill/drop explaining **why** you chose that method.

```python
# Handle missing values here
df['Units Sold'] = df['Units Sold'].fillna(df['Units Sold'].median())
df['Price Per Unit'] = df['Price Per Unit'].fillna(df['Price Per Unit'].median())
df['Customer Ratings'] = df['Customer Ratings'].fillna(df['Customer Ratings'].median())
df = df.dropna(subset=['Customer Name'])
```

```python
df['City'] = df['City'].fillna(df['City'].mode()[0])
df['Payment Method'] = df['Payment Method'].fillna(df['Payment Method'].mode()[0])
```

```python
df.isnull().sum()
```

### 3.3 Fix data types
Convert `Order Date` to datetime.  
Optionally convert suitable columns to `category` (e.g. Brand, Payment Method).

```python
# Convert Order Date (and any other types you want to fix)
df['Order Date'] = pd.to_datetime(df['Order Date'], format='mixed')
df.dtypes
```

### 3.4 Handle duplicates

```python
# Count duplicates, then remove them

df.duplicated().sum()
df = df.drop_duplicates()
```

### 3.5 Verify cleaning
Confirm missing values are handled and duplicates are gone.

```python
# Verify: isnull().sum() and duplicated().sum()
df.isnull().sum()
df.duplicated().sum()
```

## Section 5: Visualisations (3 charts)
Create **exactly three** charts. Each must have a **title** and **axis labels**.

1. **Histogram** — distribution of a numeric column (e.g. Price Per Unit, Total, Customer Age)
2. **Boxplot** — one numeric column, optionally by a category (e.g. Total by Brand)
3. **Bar chart** — counts or averages by category (e.g. Brand counts, or mean Total by Payment Method)

You may use Pandas `.plot` / `.hist` / `.boxplot` or Matplotlib / Seaborn.

```python
# Chart 1: Histogram
plt.figure(figsize=(8, 5))
plt.hist(df['Total'], bins=30)
plt.title('Distribution of Total Sales per Order')
plt.xlabel('Total (order amount)')
plt.ylabel('Number of orders')
plt.show()
```

```python
# Chart 2: Boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(x='Brand', y='Total', data=df)
plt.title('Order Total by Brand')
plt.xlabel('Brand')
plt.ylabel('Total (order amount)')
plt.show()
```

```python
# Chart 3: Bar chart
city_sales = df.groupby('City')['Total'].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 5))
city_sales.plot(kind='bar')
plt.title('Top 10 Cities by Total Sales')
plt.xlabel('City')
plt.ylabel('Total sales')
plt.xticks(rotation=45)
plt.show()
```
